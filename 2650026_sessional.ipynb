# Q1

```python
class BST_Node:
    def __init__(self):
        self.lchild = None
        self.rchild = None
        self.value = None
        self.parent = None
```

```python
def BST_Insert(value, root_node:BST_Node | None = None) -> BST_Node:
    assert value != None, "Value cannot be None"
    y = None
    x = root_node
    global iterations
    while x is not None:
        iterations += 1
        y = x
        if value < x.value: # type: ignore
            x = x.lchild # type: ignore
        else:
            x = x.rchild # type: ignore
    new_node = BST_Node()
    new_node.value = value
    new_node.parent = y # type: ignore
    if y is None:
        root_node = new_node
    elif value < y.value: # type: ignore
        y.lchild = new_node # type: ignore
    else:
        y.rchild = new_node # type: ignore
    return root_node # type: ignore
```

```python
def BST_inorder(root_node:BST_Node | None = None):
    global iterations
    if root_node is None:
        return
    iterations += 1 # type: ignore
    BST_inorder(root_node.lchild)
    print(root_node.value, end=',')
    BST_inorder(root_node.rchild)

def BST_preorder(root_node:BST_Node | None = None):
    global iterations
    if root_node is None:
        return
    iterations += 1 # type: ignore
    print(root_node.value, end=',')
    BST_preorder(root_node.lchild)
    BST_preorder(root_node.rchild)

def BST_postorder(root_node:BST_Node | None = None):
    global iterations
    if root_node is None:
        return
    iterations += 1 # type: ignore
    BST_postorder(root_node.lchild)
    BST_postorder(root_node.rchild)
    print(root_node.value, end=',')
```

```python
def BST_maximum(root_node:BST_Node | None = None) -> BST_Node | None:
    global iterations
    if root_node is None:
        return None
    while root_node.rchild is not None:
        iterations += 1 # type: ignore
        root_node = root_node.rchild
    return root_node
```

```python
def BST_Search(value, root_node:BST_Node | None = None) -> BST_Node | None:
    global iterations
    if root_node is None:
        return None
    iterations += 1 # type: ignore
    if value == root_node.value:
        return root_node
    elif value < root_node.value:
        return BST_Search(value, root_node.lchild)
    else:
        return BST_Search(value, root_node.rchild)
```

## Answer

```python
BSTroot = None
arr = [40, 20, 60, 10, 30, 50, 70]

def solution_1(root_node:BST_Node, query_a, query_b):
    global iterations
    iterations = 0
    node_a = BST_Search(query_a, root_node)
    if node_a is None:
        print(f"Search {query_a} not found, {iterations} nodes visited")
    else:
        print(f"Search {query_a} found, {iterations} nodes visited")
    iterations = 0
    node_b = BST_Search(query_b, root_node)
    if node_b is None:
        print(f"Search {query_b} not found, {iterations} nodes visited")
    else:
        print(f"Search {query_b} found, {iterations} nodes visited")

for i in arr:
    BSTroot = BST_Insert(i, BSTroot)

solution_1(BSTroot, 50, 35)
print(f"Maximum: {BST_maximum(BSTroot).value}")
```

```python
keys = [12, 7, 18, 3, 9]
heap = []

# YE MAIN CLRS SE PADHA THA YAAD HAI!!!
def lchild(i):
    return 2 * i + 1

def rchild(i):
    return 2 * i + 2

def parent(i):
    return (i - 1) // 2

# INSERTION 2650026
for k in keys:
    heap.append(k)
    i = len(heap) - 1
    while i > 0 and heap[i] < heap[parent(i)]:
        p = parent(i)
        heap[i], heap[p] = heap[p], heap[i]
        i = p

print("Heap:", heap)
print("Minimum:", heap[0])

# EXTRACTMIN 2650026
heap[0] = heap[-1] # Shortcut
heap.pop()
i, n = 0, len(heap)
while True:
    l, r, s = lchild(i), rchild(i), i
    if l < n and heap[l] < heap[s]:
        s = l
    if r < n and heap[r] < heap[s]:
        s = r
    if s == i:
        break
    heap[i], heap[s] = heap[s], heap[i]
    i = s

print("After extract-min:", heap)
```

```python
N_values = [13, 8, 21]

for N in N_values:
    degrees = []
    d = 0
    n = N
    while n > 0:
        if n & 1:          # bit d is 1 -> one tree of degree d
            degrees.append(d)
        n >>= 1
        d += 1
    print(f"N={N}: {len(degrees)} trees, degrees {' '.join(map(str, degrees))}")
```
