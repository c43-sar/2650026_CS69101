# AVL Tree

```python
# Node for AVL Tree node
class AVL_node:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None
        self.height = 1
        self.balance_factor = 0
```

```python
def get_height(node: AVL_node) -> int:
    if not node:
        return 0
    return node.height

def update_height(node: AVL_node) -> None:
    if node:
        node.height = 1 + max(get_height(node.left), get_height(node.right))

def get_balance_factor(node: AVL_node) -> int:
    if not node:
        return 0
    return get_height(node.left) - get_height(node.right)

def update_balance_factor(node: AVL_node) -> None:
    if node:
        node.balance_factor = get_balance_factor(node)
```

```python
def avl_max(node: AVL_node) -> AVL_node:
    current = node
    while current.right is not None:
        current = current.right
    return current

def avl_min(node: AVL_node) -> AVL_node:
    current = node
    while current.left is not None:
        current = current.left
    return current

def avl_successor(node: AVL_node) -> AVL_node:
    if node.right is not None:
        return avl_min(node.right)
    current = node
    while current.parent is not None and current == current.parent.right:
        current = current.parent
    return current.parent

def avl_predecessor(node: AVL_node) -> AVL_node:
    if node.left is not None:
        return avl_max(node.left)
    current = node
    while current.parent is not None and current == current.parent.left:
        current = current.parent
    return current.parent

def avl_inorder_traversal(node: AVL_node) -> list:
    result = []
    if node:
        result.extend(avl_inorder_traversal(node.left))
        result.append(node.key)
        result.extend(avl_inorder_traversal(node.right))
    return result

def avl_preorder_traversal(node: AVL_node) -> list:
    result = []
    if node:
        result.append(node.key)
        result.extend(avl_preorder_traversal(node.left))
        result.extend(avl_preorder_traversal(node.right))
    return result

def avl_postorder_traversal(node: AVL_node) -> list:
    result = []
    if node:
        result.extend(avl_postorder_traversal(node.left))
        result.extend(avl_postorder_traversal(node.right))
        result.append(node.key)
    return result

def avl_level_order_traversal(node: AVL_node) -> list:
    result = []
    if not node:
        return result
    queue = [node]
    while queue:
        current = queue.pop(0)
        result.append(current.key)
        if current.left:
            queue.append(current.left)
        if current.right:
            queue.append(current.right)
    return result
```

```python
def avl_rotation_LL(node: AVL_node) -> AVL_node:
    new_root = node.left
    node.left = new_root.right
    new_root.right = node
    update_height(node)
    update_balance_factor(node)
    update_height(new_root)
    update_balance_factor(new_root)
    return new_root

def avl_rotation_RR(node: AVL_node) -> AVL_node:
    new_root = node.right
    node.right = new_root.left
    new_root.left = node
    update_height(node)
    update_balance_factor(node)
    update_height(new_root)
    update_balance_factor(new_root)
    return new_root

def avl_rotation_LR(node: AVL_node) -> AVL_node:
    node.left = avl_rotation_RR(node.left)
    return avl_rotation_LL(node)

def avl_rotation_RL(node: AVL_node) -> AVL_node:
    node.right = avl_rotation_LL(node.right)
    return avl_rotation_RR(node)
```

```python
def avl_insert(root: AVL_node, key: int) -> AVL_node:
    # Perform standard BST insert
    if not root:
        return AVL_node(key)
    elif key < root.key:
        root.left = avl_insert(root.left, key)
    else:
        root.right = avl_insert(root.right, key)

    # Update height and balance factor of this ancestor node
    update_height(root)
    update_balance_factor(root)

    # Get the balance factor to check whether this node became unbalanced
    balance = root.balance_factor

    # If unbalanced, then there are 4 cases

    # Left Left Case
    if balance > 1 and key < root.left.key:
        return avl_rotation_LL(root)
    # Right Right Case
    if balance < -1 and key > root.right.key:
        return avl_rotation_RR(root)
    # Left Right Case
    if balance > 1 and key > root.left.key:
        return avl_rotation_LR(root)
    # Right Left Case
    if balance < -1 and key < root.right.key:
        return avl_rotation_RL(root)
    return root
```

```python
def avl_delete(root: AVL_node, key: int) -> AVL_node:
    # Perform standard BST delete
    if not root:
        return root
    elif key < root.key:
        root.left = avl_delete(root.left, key)
    elif key > root.key:
        root.right = avl_delete(root.right, key)
    else:
        # Node with only one child or no child
        if not root.left:
            return root.right
        elif not root.right:
            return root.left
        # Node with two children: Get the inorder successor (smallest in the right subtree)
        temp = avl_min(root.right)
        root.key = temp.key
        root.right = avl_delete(root.right, temp.key)

    # Update height and balance factor of this ancestor node
    update_height(root)
    update_balance_factor(root)

    # Get the balance factor to check whether this node became unbalanced
    balance = root.balance_factor

    # If unbalanced, then there are 4 cases

    # Left Left Case
    if balance > 1 and get_balance_factor(root.left) >= 0:
        return avl_rotation_LL(root)
    # Right Right Case
    if balance < -1 and get_balance_factor(root.right) <= 0:
        return avl_rotation_RR(root)
    # Left Right Case
    if balance > 1 and get_balance_factor(root.left) < 0:
        return avl_rotation_LR(root)
    # Right Left Case
    if balance < -1 and get_balance_factor(root.right) > 0:
        return avl_rotation_RL(root)
    return root
```
